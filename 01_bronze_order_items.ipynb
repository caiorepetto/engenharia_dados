Carga arquivos camada bronze

```python
#Ler arquivo

df_customers = (
    spark.read
.option("header" , "true")
.option("inferSchema", "true")
.csv("/Volumes/workspace/raw/olist/olist_customers_dataset.csv")
)

```

```python
#Verificar arquivo lido

display(df_customers)
```

```python
# Ver Schema

df_customers.printSchema()
```

```python
#Adicionar metadados de ingestão

from pyspark.sql.functions import current_timestamp, from_utc_timestamp

df_customers_bronze = (
    df_customers
    .withColumn(
        "data_ingestao",
        from_utc_timestamp(
            current_timestamp(),
            "America/Sao_Paulo"
        )
    )
)
```

```python
#Analisar df_customers_bronze

display(df_customers_bronze)
```

#Gravar em Delta

```python
(
    df_customers_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.bronze.customers")
)
```

```python
%sql
--Conferir tabela

select
*
from
workspace.bronze.customers
```
