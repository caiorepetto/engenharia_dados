```python
#Ler arquivo

df_order_payments = (
    spark.read
    .option("header","true")
    .option("inferSchema","true")
    .csv("/Volumes/workspace/raw/olist/olist_order_payments_dataset.csv")
)
```

```python
#Inspecionar dados

display(df_order_payments)
```

```python
#Verificar schema

df_order_payments.printSchema()
```

```python
#Adicionar metadados de ingestão

from pyspark.sql.functions import current_timestamp, from_utc_timestamp

df_order_payments_bronze = (
    df_order_payments
    .withColumn(
        "data_ingestao",
        from_utc_timestamp(
            current_timestamp(),
            "America/Sao_Paulo"
        )
    )
)
```

```python
#Salvar como tabela Delta

(
    df_order_payments_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.bronze.order_payments")
)
```

```python
%sql
--Verificar dados

select
*
from
workspace.bronze.order_payments
```
