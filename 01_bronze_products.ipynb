```python
#Ler arquivo

df_products = (
    spark.read
    .option("header","true")
    .option("inferSchema","true")
    .csv("/Volumes/workspace/raw/olist/olist_products_dataset.csv")
)
```

```python
#Inspecionar dados

display(df_products)
```

```python
#Verificar schema

df_products.printSchema()
```

```python
#Adicionar metadados de ingestão

from pyspark.sql.functions import current_timestamp, from_utc_timestamp

df_products_bronze = (
    df_products
    .withColumn(
        "data_ingestao",
        from_utc_timestamp(
            current_timestamp(),
            "America/Sao_Paulo"
        )
    )
)
```

```python
#Salvar como tabela Delta

(
    df_products_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.bronze.products")
)
```

```python
%sql
--Verificar dados

select
*
from
workspace.bronze.products
```
